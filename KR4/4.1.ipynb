# Задание 4.1

Корзина содержит 23 шара: 8 белых, 6 синих и 9 красных. На каждом шаге из корзины «наудачу» извлекается шар и назад в корзину не возвращается. Исход $n$ последовательных извлечений называется выборкой объёма $n$ **без возвращения** или **бесповторной выборкой**. Из корзины наудачу (без возвращения) извлечены 6 шаров. Найдите вероятности следующих событий и покажите их статистическую устойчивость:

- **а)** $A=\{\text{все шары красные}\}$;
- **б)** $B=\{3\ \text{синих},\ 2\ \text{белых и}\ 1\ \text{красный}\}$;
- **в)** $D=\{\text{в точности 4 белых шара}\}$;
- **г)** Разобрать случай извлечения шаров **с возвращением**. Найдите вероятность $\mathbb P(A)$ события $A$, если «статистическая вероятность» равна $\hat p(A)\approx 0{,}00358549$.

**Ответ:** а) $\mathbb P(A)=\dfrac{4}{4807}\approx0{,}0008321198252548367$.

*Общее требование листка:* ко всем задачам, где есть числовой ответ, напишите программу (код) с использованием инструментария **Jupyter Notebook**, который иллюстрирует график зависимости относительной частоты $\hat p(A)\overset{\text{def}}{=}\dfrac{N(A)}{N}$ события $A$ от числа проведённых реализаций опыта $N$.

## Теория

**Классическое определение вероятности.** Если пространство элементарных исходов $\Omega$ конечно и все исходы равновозможны, то

$$
\mathbb P(A)=\frac{|A|}{|\Omega|}.
$$

**Выборка без возвращения.** Порядок извлечения не важен, поэтому исход — это 6-элементное подмножество из 23 шаров:

$$
|\Omega|=C_{23}^{6}=\frac{23!}{6!\,17!}=100\,947 .
$$

Число исходов, в которых взято $k_1$ белых, $k_2$ синих и $k_3$ красных шаров ($k_1+k_2+k_3=6$), равно $C_8^{k_1}C_6^{k_2}C_9^{k_3}$ (**гипергеометрическое распределение**):

$$
\mathbb P(k_1,k_2,k_3)=\frac{C_8^{k_1}\,C_6^{k_2}\,C_9^{k_3}}{C_{23}^{6}} .
$$

- **а)** $\mathbb P(A)=\dfrac{C_9^6}{C_{23}^6}=\dfrac{84}{100\,947}=\dfrac{4}{4807}\approx0{,}000832120$;
- **б)** $\mathbb P(B)=\dfrac{C_6^3\,C_8^2\,C_9^1}{C_{23}^6}=\dfrac{20\cdot28\cdot9}{100\,947}=\dfrac{5040}{100\,947}=\dfrac{240}{4807}\approx0{,}049927$;
- **в)** $\mathbb P(D)=\dfrac{C_8^4\,C_{15}^2}{C_{23}^6}=\dfrac{70\cdot105}{100\,947}=\dfrac{7350}{100\,947}=\dfrac{350}{4807}\approx0{,}072810$ (остальные 2 шара — из 15 небелых).

**г) Выборка с возвращением.** Состав корзины не меняется, извлечения независимы, на каждом шаге $\mathbb P(\text{белый})=\frac{8}{23}$, $\mathbb P(\text{синий})=\frac{6}{23}$, $\mathbb P(\text{красный})=\frac{9}{23}$. Число шаров каждого цвета имеет **полиномиальное распределение**:

$$
\mathbb P(k_1,k_2,k_3)=\frac{6!}{k_1!\,k_2!\,k_3!}\Big(\frac{8}{23}\Big)^{k_1}\Big(\frac{6}{23}\Big)^{k_2}\Big(\frac{9}{23}\Big)^{k_3}.
$$

- $\mathbb P(A)=\Big(\dfrac{9}{23}\Big)^6=\dfrac{531\,441}{148\,035\,889}\approx0{,}0035899$ — согласуется со «статистической вероятностью» $\hat p(A)\approx0{,}00358549$;
- $\mathbb P(B)=\dfrac{6!}{3!\,2!\,1!}\cdot\dfrac{6^3\cdot8^2\cdot9}{23^6}=\dfrac{7\,464\,960}{148\,035\,889}\approx0{,}050427$;
- $\mathbb P(D)=C_6^4\Big(\dfrac{8}{23}\Big)^4\Big(\dfrac{15}{23}\Big)^2=\dfrac{13\,824\,000}{148\,035\,889}\approx0{,}093383$ (биномиальное распределение).

**Статистическая устойчивость.** Повторяем опыт $N$ раз независимо, $N(A)$ — число опытов, в которых произошло $A$. По закону больших чисел (теорема Бернулли)

$$
\hat p(A)=\frac{N(A)}{N}\xrightarrow[N\to\infty]{}\mathbb P(A),\qquad \sigma\big(\hat p(A)\big)=\sqrt{\frac{\mathbb P(A)\,(1-\mathbb P(A))}{N}} .
$$

На графике $\hat p(A)$ от $N$ видно, что частота колеблется около точной вероятности и колебания затухают как $1/\sqrt N$.

```python
import numpy as np
import matplotlib.pyplot as plt
from fractions import Fraction
from math import comb, factorial

rng = np.random.default_rng(2026)

def plot_freq(ax, hits, p, title, cond=None, skip=None):
    """График относительной частоты p̂ = N(A)/N от числа опытов N (красная линия — точная вероятность p).
    Если задано cond — строится условная частота N(A∩B)/N(B)."""
    hits = np.asarray(hits, dtype=bool)
    n = np.arange(1, hits.size + 1)
    if cond is None:
        freq = np.cumsum(hits) / n                                   # p̂(A) после каждого опыта
    else:
        nb = np.cumsum(cond)
        freq = np.where(nb > 0, np.cumsum(hits & cond) / np.maximum(nb, 1), np.nan)  # p̂(A|B)
    step = max(1, hits.size // 20000)                                # прореживаем точки для отрисовки
    ax.plot(n[::step], freq[::step], color="blue", lw=1, label=r"$\hat p$")
    ax.axhline(p, color="red", lw=1.2, label=f"P = {p:.6f}")
    skip = skip or max(50, hits.size // 300)                         # первые опыты — сильный разброс
    lo, hi = np.nanmin(freq[skip:]), np.nanmax(freq[skip:])
    lo, hi = min(lo, p), max(hi, p)
    pad = 0.15 * (hi - lo) + 1e-12
    ax.set_ylim(lo - pad, hi + pad)
    ax.set_xlim(0, hits.size)
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("N")
    ax.set_ylabel(r"$\hat p$")
    ax.grid(ls=":", color="c")
    ax.legend(fontsize=8, loc="upper right")
    return freq[-1]


# ---------- Точные вероятности ----------
W, Bl, R, n = 8, 6, 9, 6
tot = W + Bl + R
# без возвращения: гипергеометрическое распределение
exact_wo = {
    "A": Fraction(comb(R, 6), comb(tot, n)),
    "B": Fraction(comb(Bl, 3) * comb(W, 2) * comb(R, 1), comb(tot, n)),
    "D": Fraction(comb(W, 4) * comb(tot - W, 2), comb(tot, n)),
}
# с возвращением: полиномиальное распределение
pw, pb, pr = Fraction(W, tot), Fraction(Bl, tot), Fraction(R, tot)
exact_w = {
    "A": pr**6,
    "B": Fraction(factorial(6), factorial(3) * factorial(2) * factorial(1)) * pb**3 * pw**2 * pr,
    "D": comb(6, 4) * pw**4 * (1 - pw)**2,
}

# ---------- Моделирование опыта ----------
N = 1_000_000
balls = np.array([0]*W + [1]*Bl + [2]*R)          # 0 — белый, 1 — синий, 2 — красный


def counts(drawn):
    """Число белых, синих и красных шаров в каждой выборке (строке)."""
    return [(drawn == c).sum(axis=1) for c in range(3)]


# без возвращения: случайная перестановка корзины, берём первые 6 шаров (порциями, чтобы не занимать много памяти)
cw = []
for _ in range(N // 100_000):
    keys = rng.random((100_000, tot))
    idx = np.argpartition(keys, n, axis=1)[:, :n]  # индексы 6 шаров с наименьшими ключами = случайная выборка
    cw.append(np.column_stack(counts(balls[idx])))
kw, kb, kr = np.vstack(cw).T

# с возвращением: каждый из 6 шаров выбирается независимо из всех 23
drawn_r = balls[rng.integers(0, tot, size=(N, n))]
rw, rb, rr = counts(drawn_r)

events = {
    "без возвращения": {"A": kr == 6, "B": (kb == 3) & (kw == 2) & (kr == 1), "D": kw == 4},
    "с возвращением":  {"A": rr == 6, "B": (rb == 3) & (rw == 2) & (rr == 1), "D": rw == 4},
}
exact = {"без возвращения": exact_wo, "с возвращением": exact_w}
names = {"A": "A = {все красные}", "B": "B = {3 синих, 2 белых, 1 красный}", "D": "D = {ровно 4 белых}"}

# ---------- Вывод и графики ----------
fig, axes = plt.subplots(2, 3, figsize=(17, 9))
for row, mode in enumerate(events):
    print(f"--- Выборка {mode} (N = {N}) ---")
    for col, ev in enumerate(["A", "B", "D"]):
        P = exact[mode][ev]
        hits = events[mode][ev]
        ph = plot_freq(axes[row, col], hits, float(P), f"{mode}: {names[ev]}")
        print(f"  P({ev}) = {P} = {float(P):.10f};   p̂({ev}) = {ph:.8f};   |p̂ - P| = {abs(ph - float(P)):.2e}")
fig.suptitle("Задание 4.1: относительная частота события в зависимости от числа опытов N", fontsize=13)
fig.tight_layout()
plt.show()

print(f"\nг) P(A) с возвращением = (9/23)^6 = {float(exact_w['A']):.8f};  статистическая вероятность из условия 0.00358549")
```
