# Задание 2.8

Пусть $\lambda$ — комплексный корень из $1$ степени $14$, такой, что его мнимая часть больше $0$, а действительная часть максимальна среди подобных корней с положительной мнимой частью. Даны матрицы:

$$
I=\begin{pmatrix}1&0\\0&1\end{pmatrix},\qquad
A=\begin{pmatrix}7&5\\6&2\end{pmatrix}.
$$

Найдите определитель матрицы $A+\lambda I$ и укажите в ответе: 1) действительную часть этого определителя; 2) его мнимую часть.

## Теория

Корни $14$-й степени из единицы:

$$
\lambda_k=e^{2\pi ik/14}=\cos\frac{\pi k}{7}+i\sin\frac{\pi k}{7},\qquad k=0,1,\dots,13 .
$$

Они лежат на единичной окружности в вершинах правильного $14$-угольника. Положительную мнимую часть имеют корни с $k=1,\dots,6$; действительная часть $\cos\frac{\pi k}{7}$ убывает с ростом $k$, значит, она максимальна при $k=1$:

$$
\lambda=e^{i\pi/7}.
$$

Определитель:

$$
\det(A+\lambda I)=\begin{vmatrix}7+\lambda&5\\6&2+\lambda\end{vmatrix}=(7+\lambda)(2+\lambda)-30=\lambda^2+9\lambda-16 .
$$

Так как $\lambda^2=e^{2\pi i/7}$, получаем

$$
\operatorname{Re}=\cos\frac{2\pi}{7}+9\cos\frac{\pi}{7}-16,\qquad
\operatorname{Im}=\sin\frac{2\pi}{7}+9\sin\frac{\pi}{7}.
$$

```python
import numpy as np
import sympy as sp

n = 14

# Все корни 14-й степени из 1: lambda_k = exp(2πik/14), k = 0..13
ks = np.arange(n)
all_roots = np.exp(2j*np.pi*ks/n)

# Оставляем корни с Im > 0 и выбираем среди них корень с наибольшей действительной частью
upper = ks[all_roots.imag > 1e-12]
k_best = int(upper[np.argmax(all_roots[upper].real)])
print("Корни с Im > 0: k =", upper.tolist(), " ->  выбран k =", k_best)

# Точное значение lambda
lam = sp.exp(2*sp.pi*sp.I*k_best/n)
print("lambda =", lam, "=", sp.N(lam, 12))

A = sp.Matrix([[7, 5],
               [6, 2]])
I = sp.eye(2)

# Определитель матрицы A + lambda·I
det = (A + lam*I).det()
re, im = sp.simplify(sp.re(det)), sp.simplify(sp.im(det))

print("\ndet(A + λI) =", sp.expand(det))
print("1) Re det =", re, "=", sp.N(re, 12))
print("2) Im det =", im, "=", sp.N(im, 12))

# Числовая проверка через numpy
lam_num = all_roots[k_best]
print("\nПроверка (numpy):", np.linalg.det(np.array([[7, 5], [6, 2]]) + lam_num*np.eye(2)))
```
