# Задание 2.3

Вычислите (над полем $\mathbb{C}$) и изобразите на комплексной плоскости:

$$
a)\ \sqrt[8]{1};\qquad
b)\ \sqrt[4]{-\frac{18}{1+i\sqrt3}};\qquad
c)\ \sqrt[3]{\frac{1-5i}{1+i}-5\,\frac{1+2i}{2-i}+11};\qquad
d)\ \sqrt[9]{1-i}.
$$

## Теория

Над полем $\mathbb{C}$ корень $n$-й степени из $w=\rho e^{i\varphi}\neq0$ — это множество из $n$ чисел

$$
\sqrt[n]{w}=\left\{\sqrt[n]{\rho}\,\exp\!\left(i\,\frac{\varphi+2\pi k}{n}\right),\ k=0,\dots,n-1\right\},
$$

которые являются вершинами правильного $n$-угольника на окружности радиуса $\sqrt[n]{\rho}$.

**Подкоренные выражения:**

- **a)** $1=e^{i\cdot0}$ $\Rightarrow$ $z_k=e^{i\pi k/4}$: $\ \pm1,\ \pm i,\ \dfrac{\pm1\pm i}{\sqrt2}$.
- **b)** $-\dfrac{18}{1+i\sqrt3}=-\dfrac{18(1-i\sqrt3)}{4}=9\left(-\dfrac12+\dfrac{\sqrt3}{2}i\right)=9e^{2\pi i/3}$ $\Rightarrow$ $z_k=\sqrt3\,e^{i(\pi/6+\pi k/2)}$: $\ \pm\left(\dfrac32+\dfrac{\sqrt3}{2}i\right),\ \pm\left(-\dfrac{\sqrt3}{2}+\dfrac32 i\right)$.
- **c)** $\dfrac{1-5i}{1+i}=-2-3i$, $\ \dfrac{1+2i}{2-i}=i$ $\Rightarrow$ под корнем $-2-3i-5i+11=9-8i=\sqrt{145}\,e^{-i\operatorname{arctg}(8/9)}$ $\Rightarrow$ $z_k=\sqrt[6]{145}\,\exp\!\left(i\,\dfrac{-\operatorname{arctg}\frac89+2\pi k}{3}\right)$.
- **d)** $1-i=\sqrt2\,e^{-i\pi/4}$ $\Rightarrow$ $z_k=\sqrt[18]{2}\,\exp\!\left(i\left(-\dfrac{\pi}{36}+\dfrac{2\pi k}{9}\right)\right)$.

```python
import numpy as np
import matplotlib.pyplot as plt


def nth_roots(w, n):
    """Все n значений корня n-й степени из комплексного числа w."""
    k = np.arange(n)
    return abs(w)**(1/n) * np.exp(1j*(np.angle(w) + 2*np.pi*k)/n)


# (пункт, подпись, подкоренное выражение, степень корня)
tasks = [
    ("a", r"$\sqrt[8]{1}$", 1 + 0j, 8),
    ("b", r"$\sqrt[4]{-\frac{18}{1+i\sqrt{3}}}$", -18/(1 + 1j*np.sqrt(3)), 4),
    ("c", r"$\sqrt[3]{\frac{1-5i}{1+i}-5\,\frac{1+2i}{2-i}+11}$", (1 - 5j)/(1 + 1j) - 5*(1 + 2j)/(2 - 1j) + 11, 3),
    ("d", r"$\sqrt[9]{1-i}$", 1 - 1j, 9),
]

fig, axes = plt.subplots(2, 2, figsize=(13, 13))
t = np.linspace(0, 2*np.pi, 400)

for ax, (name, label, w, n) in zip(axes.flat, tasks):
    # Корни по формуле Муавра
    z = nth_roots(w, n)
    r = abs(z[0])

    print(f"{name}) подкоренное выражение w = {w:.6g},  |w| = {abs(w):.6f},  arg w = {np.angle(w):.6f}")
    for k, zk in enumerate(z):
        print(f"   z_{k} = {round(zk.real, 9) + 0:+.6f} {round(zk.imag, 9) + 0:+.6f}i")
    # Проверка: z_k^n = w
    print(f"   max |z_k^{n} - w| = {np.max(np.abs(z**n - w)):.2e}\n")

    # Рисунок: окружность радиуса |w|^(1/n), правильный n-угольник и его вершины-корни
    ax.plot(r*np.cos(t), r*np.sin(t), 'k--', lw=0.8)
    ax.plot(np.append(z.real, z.real[0]), np.append(z.imag, z.imag[0]), color='tab:blue', lw=1)
    ax.scatter(z.real, z.imag, color='red', zorder=3)
    for k, zk in enumerate(z):
        ax.annotate(f"$z_{k}$", (zk.real, zk.imag), textcoords="offset points",
                    xytext=(12*zk.real/r - 5, 12*zk.imag/r - 4), fontsize=12)

    ax.axhline(0, color='gray', lw=0.6)
    ax.axvline(0, color='gray', lw=0.6)
    ax.set_aspect('equal')
    ax.set_xlim(-1.35*r, 1.35*r)
    ax.set_ylim(-1.35*r, 1.35*r)
    ax.grid(alpha=0.3)
    ax.set_xlabel("Re z")
    ax.set_ylabel("Im z")
    ax.set_title(f"{name}) {label},  $|z_k|$ = {r:.4f}", fontsize=13)

plt.tight_layout()
plt.show()
```
