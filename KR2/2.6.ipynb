# Задание 2.6

Для многочлена

$$
p(x)=x^{11}+x^{10}+x^{8}-2x^{7}-x^{5}+x^{4}+2x^{2}+2x+2
$$

найдите все его $11$ корней: $z_1,z_2,\dots,z_{11}\in\mathbb{C}$. В ответе укажите: 1) сумму модулей $|z_1|+\dots+|z_{11}|$; 2) действительную часть корня, у которого эта действительная часть максимальна.

## Теория

По **основной теореме алгебры** многочлен степени $n$ с комплексными коэффициентами имеет ровно $n$ комплексных корней (с учётом кратности). У многочлена с действительными коэффициентами невещественные корни разбиваются на пары комплексно-сопряжённых: $z$ и $\overline z$, поэтому у многочлена нечётной степени $11$ есть хотя бы один действительный корень.

По теореме Абеля — Руффини корни многочлена степени $\geq5$ в общем случае не выражаются в радикалах, поэтому ищем их **численно**:

- `numpy.roots` находит корни как собственные значения сопровождающей матрицы многочлена;
- для контроля те же корни считаем в `sympy` (`Poly.nroots`) с повышенной точностью и подставляем их в $p(x)$.

Дополнительная проверка — формулы Виета: сумма корней равна $-a_{10}/a_{11}=-1$, произведение равно $(-1)^{11}a_0/a_{11}=-2$.

```python
import numpy as np
import sympy as sp

# Коэффициенты p(x) по убыванию степеней: x^11, x^10, ..., x, 1
coeffs = [1, 1, 0, 1, -2, 0, -1, 1, 0, 2, 2, 2]

# Все 11 комплексных корней
roots = np.roots(coeffs)

print("Корни многочлена:")
for k, zk in enumerate(sorted(roots, key=lambda r: (-r.real, r.imag)), start=1):
    print(f"  z_{k:<2} = {zk.real:+.10f} {zk.imag:+.10f}i   |z| = {abs(zk):.10f}")

# Проверка: значения p в корнях и формулы Виета
print("\nmax |p(z_k)|        =", np.max(np.abs(np.polyval(coeffs, roots))))
print("сумма корней        =", round(np.sum(roots).real, 10), "(по Виета: -1)")
print("произведение корней =", round(np.prod(roots).real, 10), "(по Виета: -2)")

# Контроль в sympy с точностью 30 знаков
x = sp.symbols('x')
roots_sp = sp.Poly(coeffs, x).nroots(n=30)
sum_abs_sp = sum(sp.Abs(r) for r in roots_sp)
max_re_sp = max(sp.re(r) for r in roots_sp)

# Ответы
print("\n1) Сумма модулей корней:           ", np.sum(np.abs(roots)))
print("   (sympy, 30 знаков):             ", sum_abs_sp)
print("2) Максимальная действительная часть:", np.max(roots.real))
print("   (sympy, 30 знаков):             ", max_re_sp)
```
