# Задание 2.1

Представьте в алгебраической, тригонометрической и показательной формах следующие комплексные выражения:

$$
a)\ \frac{(5+i)(7-6i)}{3+i};\qquad
b)\ \frac{(1+i)^5}{(1-i)^3};\qquad
c)\ \frac{(-1+i\sqrt{3})^{15}}{(1-i)^{20}}+\frac{(-1-i\sqrt{3})^{15}}{(1+i)^{20}}.
$$

## Теория

Комплексное число можно записать в трёх формах:

- **алгебраическая:** $z = x + iy$, где $x=\operatorname{Re}z$, $y=\operatorname{Im}z$;
- **тригонометрическая:** $z = r(\cos\varphi + i\sin\varphi)$, где $r=|z|=\sqrt{x^2+y^2}$, $\varphi=\arg z\in(-\pi,\pi]$;
- **показательная:** $z = re^{i\varphi}$ (формула Эйлера $e^{i\varphi}=\cos\varphi+i\sin\varphi$).

При умножении модули перемножаются, а аргументы складываются; при делении модули делятся, аргументы вычитаются. Отсюда формула Муавра: $z^n = r^n e^{in\varphi}$.

**Решение «руками»:**

- **a)** $(5+i)(7-6i)=41-23i$, $\ \dfrac{41-23i}{3+i}=\dfrac{(41-23i)(3-i)}{10}=10-11i$. Тогда $r=\sqrt{221}$, $\varphi=-\operatorname{arctg}\frac{11}{10}$.
- **b)** $1\pm i=\sqrt2\,e^{\pm i\pi/4}$, поэтому $\dfrac{(\sqrt2)^5e^{5\pi i/4}}{(\sqrt2)^3e^{-3\pi i/4}}=2e^{2\pi i}=2$.
- **c)** $-1\pm i\sqrt3=2e^{\pm 2\pi i/3}\ \Rightarrow\ (-1\pm i\sqrt3)^{15}=2^{15}e^{\pm10\pi i}=2^{15}$; $\ (1\mp i)^{20}=2^{10}e^{\mp5\pi i}=-2^{10}$. Каждое слагаемое равно $-32$, сумма равна $-64=64e^{i\pi}$.

В коде те же действия выполняет `sympy`: `expand_complex` даёт алгебраическую форму, `Abs` и `arg` — модуль и аргумент.

```python
import sympy as sp

i = sp.I

# Исходные выражения
exprs = {
    "a": (5 + i)*(7 - 6*i) / (3 + i),
    "b": (1 + i)**5 / (1 - i)**3,
    "c": (-1 + i*sp.sqrt(3))**15 / (1 - i)**20 + (-1 - i*sp.sqrt(3))**15 / (1 + i)**20,
}

for name, expr in exprs.items():
    # Алгебраическая форма: раскрываем скобки и избавляемся от i в знаменателе
    z = sp.simplify(sp.expand_complex(expr))

    # Модуль и главное значение аргумента (-pi < arg z <= pi)
    r = sp.simplify(sp.Abs(z))
    phi = sp.simplify(sp.arg(z))

    print(f"{name}) алгебраическая форма:      z = {z}")
    print(f"   модуль и аргумент:         |z| = {r},  arg z = {phi} ≈ {float(phi):.6f} рад")
    print(f"   тригонометрическая форма:  z = {r}·(cos({phi}) + i·sin({phi}))")
    print(f"   показательная форма:       z = {r}·exp(i·({phi}))")
    print()

    # Проверка: r·e^{iφ} численно совпадает с исходным выражением
    assert abs(complex(r*sp.exp(i*phi)) - complex(expr)) < 1e-9
```
