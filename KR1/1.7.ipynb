Матрица $A=(a_{ij})$ размера $n\times n$, $n=80000$.  
Диагональные элементы — арифметическая прогрессия: $a_{1,1}=0.99$, $a_{n,n}=1.01$.  
Вне диагонали отличны от нуля только $a_{1,2}=0.57$, $a_{2,1}=0.03$.

Найти: 1) $\det A$; 2) $\operatorname{tr}(A^{-1})$.


## Теория

Матрица почти диагональная, с одним $2\times2$ блоком в левом верхнем углу.  
Диагональные элементы: $d_i = 0.99 + (i-1)\cdot h$, где $h = \dfrac{1.01-0.99}{n-1}$.

$$\det A = (d_1 d_2 - a_{1,2} a_{2,1})\prod_{i=3}^{n} d_i.$$

Для следа обратной: $\operatorname{tr}(A^{-1}) = \sum_i \lambda_i^{-1}$. Для почти диагональной матрицы можно использовать блочную формулу:

$$A^{-1} = \begin{pmatrix} B^{-1} & 0 \\ 0 & D^{-1} \end{pmatrix},\quad B=\begin{pmatrix} d_1 & a_{1,2}\\ a_{2,1} & d_2\end{pmatrix},\; D=\operatorname{diag}(d_3,\dots,d_n).$$
