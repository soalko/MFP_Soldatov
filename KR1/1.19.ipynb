# Задание 1.19

При каких $\lambda$ квадратичная форма
$$Q(x_1,x_2,x_3)=\lambda x_1^2-2x_2^2-3x_3^2+2x_1x_2-2x_1x_3+2x_2x_3$$
отрицательно определена?


## Теория

Отрицательная определённость: знаки главных миноров чередуются: $\Delta_1<0,\Delta_2>0,\Delta_3<0,\dots$ (или эквивалентно матрица $-A$ положительно определена).


```python
import sympy as sp

lam = sp.symbols('lam')
A = sp.Matrix([
    [lam, 1,  -1],
    [1,  -2,  1],
    [-1,  1, -3]
])

# Отрицательная определённость <=> -A положительно определена
mA = -A
m1 = mA[0,0]
m2 = mA[:2,:2].det()
m3 = mA.det()

print("Миноры -A:")
print("Δ1 =", m1)
print("Δ2 =", sp.expand(m2))
print("Δ3 =", sp.expand(m3))

sol = sp.solve([m1>0, m2>0, m3>0], lam)
print("Решение:", sol)
```
