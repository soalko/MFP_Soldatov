# Задание 1.10

Дана матрица:
$$M=\begin{pmatrix}7&5&5&x\\2&2&6&9\\3&x&8&0\\3&1&1&0\end{pmatrix}.$$

Записать $\det M$ в виде $ax^2+bx+c$. Указать $a$ и $b$.


## Теория

Определитель символьно вычисляем через `sympy.Matrix.det()`, затем приводим к полиному по $x$ и извлекаем коэффициенты.


```python
import sympy as sp

x = sp.symbols('x')
M = sp.Matrix([
    [7, 5, 5, x],
    [2, 2, 6, 9],
    [3, x, 8, 0],
    [3, 1, 1, 0]
])

det = sp.expand(M.det())
print("det M =", det)

poly = sp.Poly(det, x)
coeffs = poly.all_coeffs()
# coeffs[0]*x^2 + coeffs[1]*x + coeffs[2]
print("a =", coeffs[0])
print("b =", coeffs[1])
print("c =", coeffs[2])
```
