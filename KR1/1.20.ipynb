# Задание 1.20

Эквивалентны ли над полем вещественных чисел квадратичные формы $Q_1$ и $Q_2$?

$$Q_1=2x_1^2+9x_2^2+3x_3^2+8x_1x_2-4x_1x_3-10x_2x_3,$$
$$Q_2=2x_1^2+3x_2^2+6x_3^2-4x_1x_2-4x_1x_3+8x_2x_3.$$


## Теория

Две квадратичные формы эквивалентны над $\mathbb{R}$ тогда и только тогда, когда их матрицы имеют одинаковую **сигнатуру** (число положительных и отрицательных собственных значений) — закон инерции Сильвестра.


```python
import numpy as np

def signature(A):
    w = np.linalg.eigvalsh(A)
    pos = np.sum(w > 1e-9)
    neg = np.sum(w < -1e-9)
    return pos, neg, w

A1 = np.array([[2, 4, -2],
               [4, 9, -5],
               [-2,-5, 3]], dtype=float)

A2 = np.array([[2, -2, -2],
               [-2, 3,  4],
               [-2, 4,  6]], dtype=float)

print("Q1: сигнатура", signature(A1))
print("Q2: сигнатура", signature(A2))

s1 = signature(A1)[:2]
s2 = signature(A2)[:2]
print("Эквивалентны?", s1 == s2)
```
