# Задание 1.2

Найти: **а)** rref расширенной матрицы и **б)** общее решение системы:

$$
\begin{cases}
5x_1 - x_2 + 5x_3 + 9x_4 - 2x_5 = -3,\\
3x_1 + 6x_3 + 9x_4 - x_5 = 4,\\
2x_1 + 4x_3 + 6x_4 - x_5 = 0.
\end{cases}
$$


## Теория

Та же схема, что и в 1.1. После приведения $[A|b]$ к RREF:

- Опорные столбцы → базисные переменные;
- Не опорные → свободные.

**Ответ (а):**
$$\text{rref}(A|b) = \begin{pmatrix}1&0&2&3&0&4\\0&1&5&6&0&7\\0&0&0&0&1&8\end{pmatrix}$$

**Ответ (б):** $x=(4-2x_3-3x_4;\;7-5x_3-6x_4;\;x_3;\;x_4;\;8)$.


```python
import sympy as sp
from sympy import Matrix

A = sp.Matrix([
    [5, -1, 5, 9, -2],
    [3,  0, 6, 9, -1],
    [2,  0, 4, 6, -1]
])
b = sp.Matrix([-3, 4, 0])

A_b = A.row_join(b)
rref, pivots = A_b.rref()
print("RREF расширенной матрицы:")
sp.pprint(rref)
print("Пивотные столбцы:", pivots)

# Свободные переменные x3, x4
x3, x4 = sp.symbols('x3 x4')
print("\nОбщее решение:")
print("x1 =", rref[0, 5] - rref[0, 2]*x3 - rref[0, 3]*x4)
print("x2 =", rref[1, 5] - rref[1, 2]*x3 - rref[1, 3]*x4)
print("x3 =", x3)
print("x4 =", x4)
print("x5 =", rref[2, 5])
```
