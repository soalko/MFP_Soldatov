# Задание 1.13

Дана матрица:
$$A=\begin{pmatrix}14&10&21\\12&15&26\\18&11&27\end{pmatrix}.$$

$Z$ — собственное значение с **наибольшей мнимой частью**. Найти $Z$ и собственный вектор $X=(1,U,V)$.  
В ответе: 1) мнимая часть $Z$; 2) действительная часть $U$.


## Теория

Собственные значения находим через `numpy.linalg.eig`.  
Из трёх значений выбираем то, у которого максимальная мнимая часть.  
Собственный вектор нормируем так, чтобы первая координата была равна 1.


```python
import numpy as np

A = np.array([[14,10,21],[12,15,26],[18,11,27]])
vals, vecs = np.linalg.eig(A)

idx = np.argmax(vals.imag)
Z = vals[idx]
X = vecs[:, idx]
X = X / X[0]   # первая координата = 1

print("Собственные значения:", vals)
print("Z =", Z)
print("Собственный вектор X =", X)
print("Im(Z) =", Z.imag)
print("Re(U) =", X[1].real)
```
