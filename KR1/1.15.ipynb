# Задание 1.15

Найти собственный ортонормированный базис и матрицу в этом базисе самосопряжённого оператора $\varphi$, заданного матрицей
$$A_{\varphi}^{(e)}=\begin{pmatrix}17&-8&4\\-8&17&-4\\4&-4&11\end{pmatrix}.$$


## Теория

Матрица симметрична → самосопряжённый оператор.  
Находим собственные значения и собственные векторы, ортогонализуем и нормируем.  
Матрица в новом базисе — диагональная с собственными значениями на диагонали.


```python
import numpy as np

A = np.array([[17,-8,4],[-8,17,-4],[4,-4,11]], dtype=float)
vals, vecs = np.linalg.eigh(A)  # для симметричных матриц

print("Собственные значения:", vals)

# Ортонормированный базис (столбцы vecs уже ортонормированы)
Q = vecs
print("Ортонормированный базис (столбцы):\n", Q)
print("Проверка Q^T Q = I:\n", Q.T @ Q)
print("Матрица в новом базисе (диагональная):\n", Q.T @ A @ Q)
```
