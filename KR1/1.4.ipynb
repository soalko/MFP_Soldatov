# Задание 1.4

Выполнить действия:

$$
\begin{pmatrix}3&0&2\\0&1&3\\2&2&0\\0&1&0\end{pmatrix}
\cdot
\begin{pmatrix}1&2&-1&2\\-2&-1&1&2\\2&1&1&2\end{pmatrix}
+
\begin{pmatrix}0&-4&6&1\\2&2&-5&-2\\2&-2&6&4\\1&3&0&1\end{pmatrix}.
$$


## Теория

Используем `sympy.Matrix` и операторы `*` (умножение) и `+` (сложение). Размеры: $4\times3 \cdot 3\times4 = 4\times4$, что совпадает с размером второй матрицы.


```python
import sympy as sp

A = sp.Matrix([
    [3, 0, 2],
    [0, 1, 3],
    [2, 2, 0],
    [0, 1, 0]
])
B = sp.Matrix([
    [ 1,  2, -1, 2],
    [-2, -1,  1, 2],
    [ 2,  1,  1, 2]
])
C = sp.Matrix([
    [0, -4,  6,  1],
    [2,  2, -5, -2],
    [2, -2,  6,  4],
    [1,  3,  0,  1]
])

result = A*B + C
print("A*B =")
sp.pprint(A*B)
print("\nA*B + C =")
sp.pprint(result)
```
