# Задание 1.18

При каких $\lambda$ квадратичная форма
$$Q(x_1,x_2,x_3)=x_1^2+4x_2^2+x_3^2+2\lambda x_1x_2+10x_1x_3+6x_2x_3$$
положительно определена?


## Теория

Критерий Сильвестра: все главные миноры матрицы квадратичной формы должны быть положительны.


```python
import sympy as sp

lam = sp.symbols('lam')
A = sp.Matrix([
    [1,   lam, 5],
    [lam, 4,   3],
    [5,   3,   1]
])

m1 = A[0,0]
m2 = A[:2,:2].det()
m3 = A.det()
print("Главные миноры:")
print("Δ1 =", m1)
print("Δ2 =", sp.expand(m2))
print("Δ3 =", sp.expand(m3))

# Условия положительной определённости: Δ1>0, Δ2>0, Δ3>0
sol = sp.solve([m1>0, m2>0, m3>0], lam)
print("Решение:", sol)
```
