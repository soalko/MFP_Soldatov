Определить тип кривой второго порядка, её каноническое уравнение и каноническую систему координат:

**а)** $5x^2+4xy+8y^2-32x-56y+80=0$;  
**б)** $5x^2+12xy-22x-12y-19=0$.


## Теория

Приводим квадратичную форму к главным осям: находим собственные значения матрицы квадратичной части, поворачиваем систему координат, затем выделяем полный квадрат по линейным членам и делаем параллельный перенос.


```python
import sympy as sp

x, y = sp.symbols('x y')

def classify(expr):
    # Квадратичная часть
    A = sp.Matrix([[sp.Rational(5), sp.Rational(2)],
                   [sp.Rational(2), sp.Rational(8)]])
    # Для общего случая нужно вычислить матрицу по expr
    return A

# --- а) ---
expr_a = 5*x**2 + 4*x*y + 8*y**2 - 32*x - 56*y + 80
A_a = sp.Matrix([[5, 2],[2, 8]])
ev_a = A_a.eigenvals()
print("а) Собственные значения квадратичной части:", ev_a)

# --- б) ---
expr_b = 5*x**2 + 12*x*y - 22*x - 12*y - 19
A_b = sp.Matrix([[5, 6],[6, 0]])
ev_b = A_b.eigenvals()
print("б) Собственные значения квадратичной части:", ev_b)

# Дополнительно: численное решение для канонического вида
import numpy as np
def canon(expr):
    # Извлекаем коэффициенты
    poly = sp.Poly(expr, x, y)
    a11 = poly.coeff_monomial(x**2)
    a12 = poly.coeff_monomial(x*y) / 2
    a22 = poly.coeff_monomial(y**2)
    b1  = poly.coeff_monomial(x)
    b2  = poly.coeff_monomial(y)
    c   = poly.coeff_monomial(1)
    A = np.array([[a11, a12],[a12, a22]], dtype=float)
    w, V = np.linalg.eigh(A)
    return w, V

w_a, V_a = canon(expr_a)
w_b, V_b = canon(expr_b)
print("а) w =", w_a, "\nV =\n", V_a)
print("б) w =", w_b, "\nV =\n", V_b)
```
