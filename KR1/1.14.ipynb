# Задание 1.14

Найти минимальный многочлен матрицы
$$A=\begin{pmatrix}
-3&7&-6&-1&8&-3&-2\\
1&-5&8&-3&-5&4&-5\\
1&3&-4&6&-2&0&8\\
-1&11&-12&9&3&-3&9\\
-3&11&-12&5&8&-4&6\\
-3&2&1&-4&4&1&-7\\
1&-5&6&-2&-4&3&-2
\end{pmatrix}.$$


## Теория

Минимальный многочлен — многочлен наименьшей степени, аннулирующий матрицу.  
Ищем его как $m(\lambda) = \lambda^k + c_{k-1}\lambda^{k-1}+\dots+c_0$, решая линейную систему $m(A)=0$ (метод неопределённых коэффициентов).


```python
import sympy as sp

A = sp.Matrix([
    [-3,  7, -6, -1,  8, -3, -2],
    [ 1, -5,  8, -3, -5,  4, -5],
    [ 1,  3, -4,  6, -2,  0,  8],
    [-1, 11,-12,  9,  3, -3,  9],
    [-3, 11,-12,  5,  8, -4,  6],
    [-3,  2,  1, -4,  4,  1, -7],
    [ 1, -5,  6, -2, -4,  3, -2]
])

lam = sp.symbols('lam')


def poly_at_matrix(poly_expr, Mat):
    """
    Аккуратно вычисляет значение полинома от матрицы:
    для каждого одночлена coeff * lam**deg берём coeff * Mat**deg.
    Так мы избегаем проблемы 'Add is not subscriptable' при subs(lam, A).
    """
    p = sp.Poly(sp.expand(poly_expr), lam)
    result = sp.zeros(*Mat.shape)
    for (deg,), coeff in p.terms():
        result += coeff * (Mat ** deg)
    return result


# 1) Характеристический многочлен и его разложение на неприводимые множители
char = A.charpoly(lam).as_expr()
print("Характеристический многочлен (в разложении):")
print(sp.factor(char))

factors = sp.factor_list(char)[1]   # [(неприводимый множитель, кратность), ...]
print("\nНеприводимые множители и их кратности в charpoly:")
for f, m in factors:
    print(f"  ({f}), кратность {m}")


# 2) Минимальный многочлен: для каждого неприводимого f находим
#    минимальную степень p ≤ m, для которой f(A)^p = 0.
minpoly = sp.Integer(1)
for f, m in factors:
    fp = sp.Integer(1)
    chosen_p = None
    for p in range(1, m + 1):
        fp = sp.expand(fp * f)             # f^p как многочлен от lam
        if poly_at_matrix(fp, A).is_zero_matrix:
            chosen_p = p
            break
    if chosen_p is None:
        # теоретически невозможно (f^m всегда аннулирует A на своей компоненте)
        chosen_p = m
    print(f"Множитель {f}: минимальная степень = {chosen_p}")
    minpoly *= fp

minpoly = sp.expand(minpoly)
print("\nМинимальный многочлен матрицы A:")
print(minpoly)
```
