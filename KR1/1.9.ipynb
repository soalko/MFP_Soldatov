# Задание 1.9

Даны матрицы:
$$I=\begin{pmatrix}1&0\\0&1\end{pmatrix},\quad A=\begin{pmatrix}7\gamma & 14\gamma\\ 9\gamma & 1\gamma\end{pmatrix},\;\gamma=0.057.$$

Пусть $S_{m,n}=\sum_{i=m}^{n}A^i$, $B_m=(I-A)^{-1}A^m$.

1) Наибольший элемент $S=S_{4,70}$.  
2) Наибольший элемент $B=B_4$.  
3) Наименьшее $k$, такое что при $n>k$ наибольший элемент $S_{k,n}<0.0002$.


## Теория

Сумма геометрической прогрессии матриц:
$$S_{m,n} = A^m(I-A^{n-m+1})(I-A)^{-1}.$$

$B_m = (I-A)^{-1}A^m$ — «хвост» суммы $\sum_{i=m}^{\infty}A^i$.

При $\|A\|<1$ ряд сходится. Здесь $\gamma=0.057$, собственные значения $A$ малы.


```python
import numpy as np

g = 0.057
I = np.eye(2)
A = np.array([[7*g, 14*g],[9*g, 1*g]])

def S(m, n):
    return sum(np.linalg.matrix_power(A, i) for i in range(m, n+1))

def B(m):
    return np.linalg.inv(I-A) @ np.linalg.matrix_power(A, m)

print("Наибольший элемент S(4,70) =", S(4,70).max())
print("Наибольший элемент B(4)    =", B(4).max())

# Поиск k
for k in range(0, 500):
    max_elem = S(k+1, k+200).max()  # n > k, берём большое n
    if max_elem < 0.0002:
        print("Наименьшее k =", k)
        break
```
