# Задание 1.3

Пусть $(x,y,z,u)$ — решение системы:

$$
\begin{cases}
32x + 97y + 90z + 48u = 47,\\
35x + 84y + 91z + 54u = 40,\\
17x + 18y + 21z + 13u = 37.
\end{cases}
$$

Объяснить, почему $y$ можно представить как $y = a + bx$; найти $a$ и $b$.


## Теория

Система из 3 уравнений с 4 неизвестными. Ранг матрицы равен 3, поэтому имеется **одна свободная переменная**. Если в качестве свободной выбрать $x$, то $y$, $z$, $u$ линейно выражаются через $x$:

$$y = a + bx.$$

Достаточно решить систему символьно (`sympy.solve`) относительно $y,z,u$, считая $x$ параметром.


```python
import sympy as sp

x = sp.symbols('x')
y, z, u = sp.symbols('y z u')

eqs = [
    32*x + 97*y + 90*z + 48*u - 47,
    35*x + 84*y + 91*z + 54*u - 40,
    17*x + 18*y + 21*z + 13*u - 37
]

sol = sp.solve(eqs, [y, z, u], dict=True)[0]
print("y =", sp.expand(sol[y]))

y_expr = sp.expand(sol[y])
a = y_expr.subs(x, 0)
b = sp.diff(y_expr, x)
print("\na =", a)
print("b =", b)
```
