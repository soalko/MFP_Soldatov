# Задание 1.8

Матрица $A=(a_{ij})$, $n\times n$, $n=43000$.  
1) Диагональные элементы — геометрическая прогрессия: $a_{1,1}=1$, $a_{n,n}=1.07$.  
2) Элементы первой строки — арифметическая прогрессия с $a_{1,n}=1.10$.  
3) Вне главной диагонали и вне первой строки — нули.

Найти $B=A^{-1}$; указать 1) наименьший элемент $B$; 2) сумму элементов $B$.


## Теория

Матрица имеет структуру:
$$A = \begin{pmatrix} a_{11} & a_{12} & \dots & a_{1n}\\ 0 & d_2 & & \\ \vdots & & \ddots & \\ 0 & & & d_n\end{pmatrix}.$$

Диагональ: $d_i = q^{i-1}$, $q = 1.07^{1/(n-1)}$.  
Первая строка: $a_{1,j} = 1 + (j-1)\cdot p$, $p = (1.10-1)/(n-1)$.  
(Предполагаем $a_{1,1}=1$ как диагональный, т.е. первая строка начинается с 1.)

Обратная матрица: верхнетреугольная, вычисляется обратной подстановкой.


```python
import numpy as np

n = 43000
q = 1.07**(1/(n-1))
p = (1.10 - 1.0)/(n-1)

# Диагональ
d = np.array([q**i for i in range(n)])

# Первая строка: a_{1,j} = 1 + (j-1)*p
row1 = 1 + np.arange(n)*p

# Матрица верхнетреугольная: A[i,i]=d[i], A[0,j]=row1[j] для j>=1
# Обратная: B[i,i] = 1/d[i]
# B[0,j] = -row1[j]/(d[0]*d[j]) для j>=1
# B[i,j] = 0 при 0<i<j

# Наименьший элемент
B_diag = 1.0/d
B_row1 = np.array([0.0] + [-row1[j]/(d[0]*d[j]) for j in range(1, n)])
min_elem = min(B_diag.min(), B_row1.min())
print("Наименьший элемент B ≈", min_elem)

# Сумма элементов
sum_B = B_diag.sum() + B_row1.sum()
print("Сумма элементов B ≈", sum_B)
```
