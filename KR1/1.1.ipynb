С использованием Jupyter Notebook найти общее решение системы линейных уравнений:

$$
\begin{cases}
x_1 + 5x_2 - 2x_3 = -7,\\
-3x_1 + x_2 + 9x_3 - 5x_4 = 9,\\
4x_1 - 8x_2 - x_3 + 7x_4 = 0.
\end{cases}
$$


## Теория

Для решения системы $Ax=b$ используем метод Гаусса — приводим расширенную матрицу $[A\,|\,b]$ к **приведённому ступенчатому виду** (RREF, Reduced Row Echelon Form) с помощью `sympy.Matrix.rref()`.

- Ведущие столбцы соответствуют базисным переменным.
- Свободные переменные выражаются через параметры.
- Общее решение записывается как $x = x_{\text{част}} + \sum t_i \cdot v_i$.


```python
import sympy as sp
from sympy import Matrix

# Матрица коэффициентов A и вектор правой части b
A = sp.Matrix([
    [ 1,  5, -2,  0],
    [-3,  1,  9, -5],
    [ 4, -8, -1,  7]
])
b = sp.Matrix([-7, 9, 0])

# Расширенная матрица [A | b]
A_b = A.row_join(b)
print("Расширенная матрица [A|b]:")
sp.pprint(A_b)

# Приводим к RREF
rref, pivots = A_b.rref()
print("\nRREF:")
sp.pprint(rref)
print("\nПивотные столбцы:", pivots)

# Свободная переменная x3 (если она не пивотная)
# Из RREF получаем: x1, x2, x4 — базисные, x3 — свободная
t = sp.symbols('t')
x1 = -11*sp.Rational(0,1)  # заполним ниже символьно
# Восстановим выражения из RREF
x1_expr = rref[0, 4] - rref[0, 2]*t
x2_expr = rref[1, 4] - rref[1, 2]*t
x4_expr = rref[2, 4] - rref[2, 2]*t

print("\nОбщее решение:")
print("x1 =", sp.simplify(x1_expr))
print("x2 =", sp.simplify(x2_expr))
print("x3 =", t)
print("x4 =", sp.simplify(x4_expr))
```
