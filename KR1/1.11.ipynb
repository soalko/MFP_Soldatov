# Задание 1.11

Используя графические возможности, изобразить матрицу $n\times n$ ($n=100$), каждая строка которой содержит биномиальные коэффициенты по модулю 2 (синий = 1, жёлтый = 0).


## Теория

Матрица $M_{i,j} = \binom{i}{j} \bmod 2$ при $0\le j\le i<n$ — это треугольник Паскаля по модулю 2 (фрактал Серпинского).  
Для визуализации используем `matplotlib.pyplot.imshow` с цветовой картой.


```python
import numpy as np
import matplotlib.pyplot as plt
from math import comb

n = 100
M = np.zeros((n, n), dtype=int)
for i in range(n):
    for j in range(i+1):
        M[i, j] = comb(i, j) % 2

plt.figure(figsize=(8, 8))
plt.imshow(M, cmap='viridis', interpolation='nearest')
plt.title("Биномиальные коэффициенты по модулю 2 (n=100)")
plt.xlabel("j")
plt.ylabel("i")
plt.show()
```
