# Задание 1.5

Найти определители порядка $n$ с элементами $a_{ij}$, заданными условием:

**а)** $a_{ij} = \min(i,j)$;  
**б)** $a_{ij} = \max(i,j)$;  
**в)** $a_{ij} = |i-j|$.


## Теория

Для каждой матрицы найдём общую формулу определителя $D_n$ через минорное разложение или приведение к треугольному виду.

- **а)** $\min(i,j)$ — миноры приводят к $D_n = 1$.
- **б)** $\max(i,j)$ — $D_n = (-1)^{n-1} n$.
- **в)** $|i-j|$ — рекуррентность $D_n = -2 D_{n-1} - D_{n-2}$, $D_n = (-1)^{n-1} 2^{n-2}(n-1)$.

Рекуррентность — это свойство процесса или объекта, при котором каждый новый шаг или элемент определяется на основе предыдущих.

```python
import sympy as sp

def det_min(n):
    return sp.Matrix(n, n, lambda i, j: min(i+1, j+1)).det()

def det_max(n):
    return sp.Matrix(n, n, lambda i, j: max(i+1, j+1)).det()

def det_abs(n):
    return sp.Matrix(n, n, lambda i, j: abs(i-j)).det()

print("n | min | max | |i-j|")
for n in range(1, 8):
    print(n, det_min(n), det_max(n), det_abs(n))

# Проверим найденные формулы символьно
n = sp.symbols('n', integer=True, positive=True)
print("\nФормулы (проверка численно):")
for n_val in range(1, 9):
    print(n_val,
          "| min =", 1,
          "| max =", (-1)**(n_val-1)*n_val,
          "| |i-j| =", (-1)**(n_val-1)*2**(n_val-2)*(n_val-1) if n_val >= 2 else 0)
```
