## Теория

**1. Постановка задачи.** Бильярдный стол — прямоугольник $\Pi=[0,a]\times[0,b]$. Шар — материальная точка с начальным положением $(x_0,y_0)\in\Pi$ и скоростью $(v_x,v_y)$. Внутри $\Pi$ сил нет:

$$
\ddot x=0,\qquad \ddot y=0\qquad\Longrightarrow\qquad x=x_0+v_xt,\quad y=y_0+v_yt\quad(\text{до первого удара}).
$$

**Граничные условия (упругое отражение).** Угол падения равен углу отражения, модуль скорости сохраняется. Меняет знак только нормальная к стенке компонента:

$$
x=0\ \text{или}\ x=a:\ \ v_x\mapsto -v_x;\qquad\qquad y=0\ \text{или}\ y=b:\ \ v_y\mapsto -v_y .
$$

Значит, во всё время движения сохраняются $|v_x|$ и $|v_y|$.

**2. Метод развёртки.** Вместо того чтобы отражать траекторию, отражаем стол: при ударе о стенку продолжаем прямую в зеркальную копию прямоугольника. Копии замощают плоскость, и шар в «развёрнутой» плоскости движется по прямой

$$
X(t)=x_0+v_xt,\qquad Y(t)=y_0+v_yt .
$$

Настоящее положение получается обратным «складыванием» копий:

$$
x(t)=F_a\big(X(t)\big),\quad y(t)=F_b\big(Y(t)\big),\qquad F_a(X)=a-\big|\,a-(X \bmod 2a)\,\big| .
$$

Здесь $F_a$ — «треугольная пила» с периодом $2a$. Каждое пересечение прямой с линией $X=ka$ (или $Y=kb$) — это один удар о вертикальную (горизонтальную) стенку. Число ударов к моменту $t$ равно $N_x=\big|\lfloor X/a\rfloor-\lfloor x_0/a\rfloor\big|$ и аналогично $N_y$.

**Движение на торе.** $F_a$ и $F_b$ периодичны с периодами $2a$ и $2b$, поэтому достаточно знать $X \bmod 2a$ и $Y \bmod 2b$. Прямоугольник $[0,2a)\times[0,2b)$ (четыре копии стола) с отождествлёнными противоположными сторонами — это **тор** $\mathbb T^2$. Бильярд эквивалентен равномерному прямолинейному движению (обмотке) на торе с углами

$$
\varphi_x=\frac{\pi X}{a}=\omega_xt+\varphi_{x0},\qquad \varphi_y=\frac{\pi Y}{b}=\omega_yt+\varphi_{y0},\qquad \omega_x=\frac{\pi|v_x|}{a},\ \ \omega_y=\frac{\pi|v_y|}{b}.
$$

**3. Периодичность.** Периоды по координатам: $T_x=\dfrac{2a}{|v_x|}$, $T_y=\dfrac{2b}{|v_y|}$. Траектория замкнута тогда и только тогда, когда найдётся $T>0$ с $T=mT_x=nT_y$ ($m,n\in\mathbb N$):

$$
\boxed{\ \frac{T_x}{T_y}=\frac{\omega_y}{\omega_x}=\frac{a\,|v_y|}{b\,|v_x|}=\frac{a}{b}\,\operatorname{tg}\alpha=\frac nm\in\mathbb Q\ }
$$

Здесь $\alpha$ — угол скорости с осью $x$. Тогда период равен $T=mT_x$, и за период шар делает $2m$ ударов о вертикальные стенки и $2n$ — о горизонтальные (при старте с $v_x\ne0$, $v_y\ne0$ не из угла).

**Эргодичность.** Если отношение $\dfrac{a|v_y|}{b|v_x|}$ иррационально, то по теореме Кронекера—Вейля обмотка тора **всюду плотна и равномерно распределена**. Для любой непрерывной функции $f$ среднее по времени равно среднему по пространству:

$$
\lim_{T\to\infty}\frac1T\int_0^T f\big(x(t),y(t)\big)\,dt=\frac{1}{ab}\iint_{\Pi} f\,dx\,dy .
$$

В частности, доля времени, проведённая шаром в области $D\subset\Pi$, стремится к $|D|/(ab)$. Эргодичность здесь имеет место на каждом инвариантном торе $\{|v_x|,|v_y|=\mathrm{const}\}$, но не во всём фазовом пространстве (там есть два интеграла).

**4. Гамильтонова формулировка.** Гамильтониан с потенциалом-«стенкой»:

$$
H=\frac{p_x^2+p_y^2}{2m}+U(x,y),\qquad U=\begin{cases}0,&(x,y)\in\Pi,\\ +\infty,&(x,y)\notin\Pi.\end{cases}
$$

Функции $F_1=p_x^2$ и $F_2=p_y^2$ сохраняются (отражения меняют только знаки импульсов) и находятся в инволюции: $\{F_1,F_2\}=0$, $H=\frac{F_1+F_2}{2m}$. Система **интегрируема по Лиувиллю**: совместная поверхность уровня $\{|p_x|=P_1,\ |p_y|=P_2\}$ — это четыре экземпляра $\Pi$ с импульсами $(\pm P_1,\pm P_2)$, склеенные по сторонам, то есть тор. Переменные действие—угол:

$$
I_x=\frac1{2\pi}\oint p_x\,dx=\frac{aP_1}{\pi},\qquad I_y=\frac{bP_2}{\pi},\qquad
H=\frac{\pi^2}{2m}\left(\frac{I_x^2}{a^2}+\frac{I_y^2}{b^2}\right),
$$

$$
\omega_x=\frac{\partial H}{\partial I_x}=\frac{\pi^2I_x}{ma^2}=\frac{\pi|v_x|}{a},\qquad \omega_y=\frac{\pi|v_y|}{b},\qquad \varphi_{x,y}(t)=\omega_{x,y}t+\varphi_{x,y}(0).
$$

**Резонансный** тор ($\omega_x/\omega_y\in\mathbb Q$) заполнен замкнутыми траекториями. **Нерезонансный** тор заполнен всюду плотной обмоткой — это и есть эргодический случай.

**Что делаем в коде.** Траекторию считаем точно по формулам развёртки. Сравниваем случаи $\dfrac{T_x}{T_y}=\dfrac32$ (периодический) и $\dfrac{T_x}{T_y}=\sqrt2$ (эргодический). Строим развёртку с нумерацией ударов и образ траекторий на торе в $\mathbb R^3$: $\big((R+r\cos\varphi_y)\cos\varphi_x,\ (R+r\cos\varphi_y)\sin\varphi_x,\ r\sin\varphi_y\big)$. Анимация встроена через `to_jshtml()` — проигрыватель с паузой, шагом и регулировкой скорости.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.patches import Rectangle
from fractions import Fraction
from IPython.display import HTML, display

a, b = 2.0, 1.0          # стороны стола
x0, y0 = 0.3, 0.45       # начальное положение шара


def velocity(ratio, speed=1.0):
    """Скорость (vx, vy) единичного модуля, для которой T_x/T_y = a|vy|/(b|vx|) = ratio."""
    ang = np.arctan(ratio * b / a)
    return speed*np.cos(ang), speed*np.sin(ang)


def fold(X, L):
    """'Складывание' развёртки: F_L(X) = L - |L - (X mod 2L)|."""
    return L - np.abs(L - np.mod(X, 2*L))


def billiard(t, vx, vy):
    """Положение шара в момент t по методу развёртки и прямая в развёрнутой плоскости."""
    X, Y = x0 + vx*t, y0 + vy*t
    return fold(X, a), fold(Y, b), X, Y


def n_hits(X, Y):
    """Число ударов о вертикальные и горизонтальные стенки = число пересечённых линий сетки."""
    return np.abs(np.floor(X/a) - np.floor(x0/a)).astype(int), np.abs(np.floor(Y/b) - np.floor(y0/b)).astype(int)


def torus(X, Y, R=2.0, r=0.75):
    """Образ точки развёртки на торе: углы φx = πX/a, φy = πY/b."""
    px, py = np.pi*X/a, np.pi*Y/b
    return (R + r*np.cos(py))*np.cos(px), (R + r*np.cos(py))*np.sin(px), r*np.sin(py)


cases = {"rat": (1.5, "tab:blue", r"$T_x/T_y=3/2$ (периодическая)"),
         "irr": (np.sqrt(2), "tab:red", r"$T_x/T_y=\sqrt{2}$ (эргодическая)")}

# ---------- Гамильтонова картина: частоты и их отношение ----------
for key, (ratio, _, _) in cases.items():
    vx, vy = velocity(ratio)
    wx, wy = np.pi*abs(vx)/a, np.pi*abs(vy)/b                  # ω = ∂H/∂I
    Tx, Ty = 2*a/abs(vx), 2*b/abs(vy)
    fr = Fraction(Tx/Ty).limit_denominator(50)
    print(f"{key}: v = ({vx:.4f}, {vy:.4f}), ω_x = {wx:.4f}, ω_y = {wy:.4f}, T_x/T_y = {Tx/Ty:.6f} ≈ {fr}")

# ---------- Проверка эргодичности: доля времени в круге D (центр (1, 0.5), радиус 0.2) ----------
cD, rD = (1.0, 0.5), 0.2
t_long = np.linspace(0, 5000, 2_000_001)
for key, (ratio, _, _) in cases.items():
    x, y, _, _ = billiard(t_long, *velocity(ratio))
    frac = np.mean((x - cD[0])**2 + (y - cD[1])**2 < rD**2)       # среднее по времени
    print(f"{key}: доля времени в D = {frac:.4f},  доля площади |D|/(ab) = {np.pi*rD**2/(a*b):.4f}")

# ---------- Статический рисунок ----------
fig = plt.figure(figsize=(15, 12.5))
ax1, ax2, ax3 = (fig.add_subplot(2, 2, k) for k in (1, 2, 3))
ax4 = fig.add_subplot(2, 2, 4, projection="3d")

# (а) периодические траектории: T_x/T_y = 1 и 3/2, рисуем ровно один период
for ratio, col, (m, n) in [(1.0, "tab:green", (1, 1)), (1.5, "tab:blue", (2, 3))]:
    vx, vy = velocity(ratio)
    T = m * 2*a/abs(vx)                                       # период T = m·T_x = n·T_y
    t = np.linspace(0, T, 4000)
    x, y, _, _ = billiard(t, vx, vy)
    ax1.plot(x, y, color=col, lw=2, label=rf"$T_x/T_y={n}/{m}$: период $T={T:.2f}$, ударов {2*m}+{2*n}")
ax1.set_title("Периодические траектории (рациональный случай)")

# (б) эргодическая траектория: T_x/T_y = √2, траектория заполняет стол
vx, vy = velocity(np.sqrt(2))
t = np.linspace(0, 150, 100_001)
x, y, _, _ = billiard(t, vx, vy)
ax2.plot(x, y, color="tab:red", lw=0.5, label=r"$T_x/T_y=\sqrt{2}$, $t\in[0,150]$")
ax2.set_title("Эргодическая траектория (иррациональный случай)")

for ax in (ax1, ax2):
    ax.add_patch(Rectangle((0, 0), a, b, fill=False, lw=2.5, color="k"))
    ax.add_patch(plt.Circle(cD, rD, fill=False, ls="--", lw=1.5, color="k", label="область $D$ (проверка)"))
    ax.plot(x0, y0, "ko", ms=8, label="начальное положение", zorder=5)
    ax.set(xlim=(-0.1, a + 0.1), ylim=(-0.1, b + 0.1), aspect="equal", xlabel="$x$", ylabel="$y$")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper center", bbox_to_anchor=(0.5, -0.13), fontsize=9, ncol=2)   # легенда под графиком

# (в) метод развёртки: прямая в замощении зеркальными копиями ↔ ломаная в исходном столе
vx, vy = velocity(1.5)
tmax = 9.0
t = np.linspace(0, tmax, 3000)
x, y, X, Y = billiard(t, vx, vy)
for i in range(5):
    for j in range(7):
        ax3.add_patch(Rectangle((i*a, j*b), a, b, fc="tab:orange" if (i + j) % 2 else "white", alpha=0.15, ec="gray"))
ax3.add_patch(Rectangle((0, 0), a, b, fc="yellow", alpha=0.6, ec="k", lw=2))
ax3.plot(X, Y, color="tab:blue", lw=2, label="прямая в развёртке $(X(t),Y(t))$")
ax3.plot(x, y, color="tab:red", lw=2.2, label="настоящая траектория (сложенная)")
# моменты пересечения линий сетки = моменты ударов; нумеруем их на прямой и на ломаной
hits = sorted([(k*a - x0)/vx for k in range(1, 6)] + [(k*b - y0)/vy for k in range(1, 8)])
hits = [th for th in hits if th < tmax]
for k, th in enumerate(hits, 1):
    xs, ys, Xs, Ys = billiard(th, vx, vy)
    ax3.plot(Xs, Ys, "o", color="tab:blue", ms=5)
    ax3.annotate(str(k), (Xs, Ys), xytext=(4, -10), textcoords="offset points", color="tab:blue", fontsize=9)
    ax3.plot(xs, ys, "o", color="tab:red", ms=5)
ax3.plot(x0, y0, "ko", ms=7, label="старт")
ax3.set(xlim=(-0.2, 8.2), ylim=(-0.2, 6.2), aspect="equal", xlabel="$X$", ylabel="$Y$",
        title=f"Метод развёртки: {len(hits)} ударов = {len(hits)} пересечений линий сетки")
ax3.legend(loc="upper center", bbox_to_anchor=(0.5, -0.1), fontsize=9, ncol=3)   # легенда под графиком

# (г) обмотка тора: замкнутая (рациональный случай) и всюду плотная (иррациональный)
pu, pv = np.meshgrid(np.linspace(0, 2*np.pi, 60), np.linspace(0, 2*np.pi, 30))
ax4.plot_surface((2 + 0.75*np.cos(pv))*np.cos(pu), (2 + 0.75*np.cos(pv))*np.sin(pu), 0.75*np.sin(pv),
                 color="lightgray", alpha=0.25, linewidth=0)
for key, (ratio, col, lab) in cases.items():
    vx, vy = velocity(ratio)
    tt = np.linspace(0, 10 if key == "rat" else 60, 20000)
    _, _, X, Y = billiard(tt, vx, vy)
    ax4.plot(*torus(X, Y), color=col, lw=1.6 if key == "rat" else 0.5, label=lab)
ax4.set(xlim=(-2.8, 2.8), ylim=(-2.8, 2.8), zlim=(-1, 1))
ax4.set_box_aspect((5.6, 5.6, 2)); ax4.view_init(elev=40, azim=-60); ax4.set_axis_off()
ax4.set_title("Эквивалентное движение на торе $[0,2a)\\times[0,2b)$")
ax4.legend(loc="upper center", bbox_to_anchor=(0.5, 0.02), fontsize=9, ncol=2)

fig.suptitle(f"Бильярд в прямоугольнике $a={a:g}$, $b={b:g}$", fontsize=14)
fig.tight_layout()
fig.savefig("3.2.png", dpi=150, bbox_inches="tight")       # сохраняем статическую картинку
plt.show()

# ---------- Анимация ----------
n_frames, t_end = 120, 24.0
t_fine = np.linspace(0, t_end, 6001)
step = (len(t_fine) - 1) // (n_frames - 1)
traj = {k: billiard(t_fine, *velocity(r)) for k, (r, _, _) in cases.items()}

fig = plt.figure(figsize=(12, 9), dpi=58)
axA, axB, axC = (fig.add_subplot(2, 2, k) for k in (1, 2, 3))
axD = fig.add_subplot(2, 2, 4, projection="3d")

art = {}
for ax, key in ((axA, "rat"), (axB, "irr")):
    ratio, col, lab = cases[key]
    ax.add_patch(Rectangle((0, 0), a, b, fill=False, lw=2.5, color="k"))
    ax.set(xlim=(-0.1, a + 0.1), ylim=(-0.1, b + 0.1), aspect="equal", xlabel="$x$", ylabel="$y$", title=lab)
    ax.grid(alpha=0.3)
    tr, = ax.plot([], [], color=col, lw=1.2 if key == "rat" else 0.6)
    ball, = ax.plot([], [], "o", color=col, mec="k", ms=11)
    txt = ax.text(0.02, 0.97, "", transform=ax.transAxes, va="top", family="monospace", fontsize=9,
                  bbox=dict(boxstyle="round", fc="white", alpha=0.85))
    art[key] = (tr, ball, txt)

# развёртка (рациональный случай) с «камерой», следящей за шаром
vxr, vyr = velocity(1.5)
Xr, Yr = traj["rat"][2], traj["rat"][3]
gi, gj = np.meshgrid(np.arange(-2, 16), np.arange(-3, 22))
axC.pcolormesh(np.arange(-2, 17)*a, np.arange(-3, 23)*b, ((gi + gj) % 2).astype(float),
               cmap="Oranges", vmin=0, vmax=4, shading="flat")
for k in range(-2, 17):
    axC.axvline(k*a, color="gray", lw=0.6)
for k in range(-3, 23):
    axC.axhline(k*b, color="gray", lw=0.6)
axC.add_patch(Rectangle((0, 0), a, b, fc="yellow", alpha=0.7, ec="k", lw=2))
cell = Rectangle((0, 0), a, b, fill=False, ec="tab:red", lw=2.5)
axC.add_patch(cell)
lineC, = axC.plot([], [], color="tab:blue", lw=2)
ballC, = axC.plot([], [], "o", color="tab:blue", mec="k", ms=10)
axC.set(aspect="equal", xlabel="$X$", ylabel="$Y$", title="Развёртка (3/2): прямая в зеркальных копиях")

# тор
pu2, pv2 = np.meshgrid(np.linspace(0, 2*np.pi, 25), np.linspace(0, 2*np.pi, 13))   # лёгкая сетка тора
axD.plot_wireframe((2 + 0.75*np.cos(pv2))*np.cos(pu2), (2 + 0.75*np.cos(pv2))*np.sin(pu2), 0.75*np.sin(pv2),
                   color="lightgray", lw=0.5)
tor = {}
for key, (_, col, lab) in cases.items():
    P = torus(traj[key][2], traj[key][3])
    tl, = axD.plot([], [], [], color=col, lw=1.2, label=lab)
    pt, = axD.plot([], [], [], "o", color=col, mec="k", ms=8)
    tor[key] = (P, tl, pt)
axD.set_box_aspect((5.6, 5.6, 2)); axD.view_init(elev=40, azim=-60); axD.set_axis_off()
axD.set(xlim=(-2.8, 2.8), ylim=(-2.8, 2.8), zlim=(-1, 1))
axD.set_title("Обмотка тора")
axD.legend(loc="upper center", bbox_to_anchor=(0.5, 0.02), fontsize=9)


def update(i):
    j = i*step
    t = t_fine[j]
    for key, (tr, ball, txt) in art.items():
        x, y, X, Y = traj[key]
        vx, vy = velocity(cases[key][0])
        tr.set_data(x[:j+1], y[:j+1])
        ball.set_data([x[j]], [y[j]])
        Nx, Ny = n_hits(X[j], Y[j])
        sx = 1 if np.mod(X[j], 2*a) < a else -1             # текущий знак v_x после отражений
        sy = 1 if np.mod(Y[j], 2*b) < b else -1
        txt.set_text(f"t = {t:5.1f}  удары: x {Nx:2d}, y {Ny:2d}\nv = ({sx*vx:+.3f}, {sy*vy:+.3f})")
    # развёртка: прямая, шар и копия стола, в которой он сейчас находится
    lineC.set_data(Xr[:j+1], Yr[:j+1])
    ballC.set_data([Xr[j]], [Yr[j]])
    cell.set_xy((np.floor(Xr[j]/a)*a, np.floor(Yr[j]/b)*b))
    axC.set_xlim(Xr[j] - 2*a, Xr[j] + 2*a)
    axC.set_ylim(Yr[j] - 2*b, Yr[j] + 2*b)
    # тор
    for key, (P, tl, pt) in tor.items():
        tl.set_data(P[0][:j+1], P[1][:j+1]); tl.set_3d_properties(P[2][:j+1])
        pt.set_data([P[0][j]], [P[1][j]]); pt.set_3d_properties([P[2][j]])
    return []


anim = FuncAnimation(fig, update, frames=n_frames, interval=60)
fig.tight_layout()
plt.close(fig)
display(HTML(anim.to_jshtml(default_mode="once")))   # проигрыватель: пуск/пауза, шаг, скорость
```
